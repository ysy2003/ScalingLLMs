# Alternative Robustness Metrics Analysis

This notebook analyzes robustness using alternative metrics:
- **Consistency**: Penalizes any change (both improvement and degradation)
- **Worst-Case**: Maximum per-sample degradation
- **Correlation**: Rank preservation between clean and perturbed scores

These metrics complement the current degradation rate formula.

```python
import json
import numpy as np
from scipy.stats import spearmanr, pearsonr
import os
from pathlib import Path

# Base directory
BASE_DIR = Path('/root/Design2code/robustness_results')
```

```python
def load_robustness_data(model_dir):
    """Load robustness data from JSON file."""
    # Try different JSON file names
    for filename in ['robustness_report.json', 'robustness_results.json']:
        json_path = model_dir / filename
        if json_path.exists():
            with open(json_path) as f:
                return json.load(f)
    return None

def print_current_report(model_dir):
    """Print the current robustness report text file."""
    txt_path = model_dir / 'robustness_report.txt'
    if txt_path.exists():
        print("=" * 60)
        print(f"CURRENT REPORT: {model_dir.name}")
        print("=" * 60)
        with open(txt_path) as f:
            print(f.read())
    else:
        print(f"No robustness_report.txt found in {model_dir}")
```

```python
def compute_consistency(clean_scores, perturbed_scores):
    """
    Consistency Score: Penalizes any change (both improvement and degradation).
    
    Formula: 1 - |clean - perturbed| / max(clean, perturbed)
    
    Range: [0, 1], higher is better
    A truly robust model should produce similar results regardless of perturbation.
    """
    clean = np.array(clean_scores)
    perturbed = np.array(perturbed_scores)
    
    abs_diff = np.abs(clean - perturbed)
    max_scores = np.maximum(clean, perturbed)
    
    # Avoid division by zero
    consistency_per_sample = np.where(
        max_scores > 0,
        1 - abs_diff / max_scores,
        1.0  # If both are 0, they're "consistent"
    )
    
    return {
        'mean': float(np.mean(consistency_per_sample)),
        'std': float(np.std(consistency_per_sample)),
        'min': float(np.min(consistency_per_sample)),
        'per_sample': consistency_per_sample
    }
```

```python
def compute_worst_case(clean_scores, perturbed_scores):
    """
    Worst-Case Metrics: Maximum per-sample degradation.
    
    For safety-critical applications, we care about the worst case.
    
    Returns:
    - max_degradation: Largest drop from clean to perturbed
    - p95_degradation: 95th percentile degradation
    - max_change: Largest absolute change (improvement or degradation)
    """
    clean = np.array(clean_scores)
    perturbed = np.array(perturbed_scores)
    
    # Degradation = clean - perturbed (positive = got worse)
    degradation = clean - perturbed
    
    # Relative degradation (as percentage of clean)
    rel_degradation = np.where(
        clean > 0,
        degradation / clean,
        0.0
    )
    
    abs_change = np.abs(degradation)
    
    return {
        'max_degradation': float(np.max(degradation)),
        'max_degradation_pct': float(np.max(rel_degradation) * 100),
        'p95_degradation': float(np.percentile(degradation, 95)),
        'p95_degradation_pct': float(np.percentile(rel_degradation, 95) * 100),
        'max_abs_change': float(np.max(abs_change)),
        'samples_degraded': int(np.sum(degradation > 0.01)),  # >1% drop
        'samples_improved': int(np.sum(degradation < -0.01)),  # >1% improvement
    }
```

```python
def compute_correlation(clean_scores, perturbed_scores):
    """
    Correlation Metrics: Do rankings stay stable?
    
    A robust model should preserve relative rankings:
    if sample A > sample B on clean, then A > B on perturbed too.
    
    Returns:
    - spearman: Spearman rank correlation
    - pearson: Pearson linear correlation
    """
    clean = np.array(clean_scores)
    perturbed = np.array(perturbed_scores)
    
    # Handle edge cases
    if len(clean) < 3 or np.std(clean) == 0 or np.std(perturbed) == 0:
        return {
            'spearman': None,
            'spearman_pvalue': None,
            'pearson': None,
            'pearson_pvalue': None
        }
    
    spearman_corr, spearman_p = spearmanr(clean, perturbed)
    pearson_corr, pearson_p = pearsonr(clean, perturbed)
    
    return {
        'spearman': float(spearman_corr),
        'spearman_pvalue': float(spearman_p),
        'pearson': float(pearson_corr),
        'pearson_pvalue': float(pearson_p)
    }
```

```python
def compute_combined_robustness(clean_scores, perturbed_scores):
    """
    Combined Robustness Score.
    
    Weights:
    - 40% Consistency (primary: consistent outputs)
    - 20% Correlation (rankings preserved)
    - 20% Mean Ratio (mean performance preserved)
    - 20% Worst-Case (no catastrophic drops)
    
    Range: [0, 1], higher is better
    """
    clean = np.array(clean_scores)
    perturbed = np.array(perturbed_scores)
    
    # 1. Consistency
    consistency = compute_consistency(clean, perturbed)['mean']
    
    # 2. Correlation (Spearman)
    corr_data = compute_correlation(clean, perturbed)
    correlation = max(0, corr_data['spearman'] if corr_data['spearman'] else 0)
    
    # 3. Mean Ratio
    clean_mean = np.mean(clean)
    perturbed_mean = np.mean(perturbed)
    if max(clean_mean, perturbed_mean) > 0:
        mean_ratio = min(clean_mean, perturbed_mean) / max(clean_mean, perturbed_mean)
    else:
        mean_ratio = 1.0
    
    # 4. Worst-case bound
    abs_diff = np.abs(clean - perturbed)
    max_scores = np.maximum(clean, perturbed)
    worst_case_ratio = np.where(max_scores > 0, abs_diff / max_scores, 0)
    worst_case = max(0, 1 - np.max(worst_case_ratio))
    
    # Combined
    combined = (
        0.4 * consistency +
        0.2 * correlation +
        0.2 * mean_ratio +
        0.2 * worst_case
    )
    
    return {
        'combined': float(combined),
        'consistency': float(consistency),
        'correlation': float(correlation),
        'mean_ratio': float(mean_ratio),
        'worst_case': float(worst_case)
    }
```

```python
def get_robustness_rating(score):
    """Get rating based on combined robustness score."""
    if score >= 0.95:
        return "EXCELLENT"
    elif score >= 0.85:
        return "GOOD"
    elif score >= 0.70:
        return "MODERATE"
    elif score >= 0.50:
        return "POOR"
    else:
        return "UNRELIABLE"
```

```python
def analyze_model(model_name, model_dir):
    """
    Full analysis of a model's robustness.
    """
    data = load_robustness_data(model_dir)
    
    if data is None:
        print(f"No data found for {model_name}")
        return None
    
    # Check if per_sample data exists
    if 'per_sample' not in data:
        print(f"\n{model_name}: No per-sample data available (only aggregate metrics)")
        print(f"  Visual Fidelity - Clean: {data.get('visual_fidelity', {}).get('combined', {}).get('clean', 'N/A'):.4f}")
        print(f"  Visual Fidelity - Perturbed: {data.get('visual_fidelity', {}).get('combined', {}).get('perturbed', 'N/A'):.4f}")
        return None
    
    per_sample = data['per_sample']
    
    results = {
        'model': model_name,
        'samples': data.get('samples', len(per_sample.get('clean_clip', []))),
        'strength': data.get('strength', 0.2)
    }
    
    # Analyze each metric type
    metrics = [
        ('CLIP', 'clean_clip', 'perturbed_clip'),
        ('IoU', 'clean_iou', 'perturbed_iou'),
        ('Structural', 'clean_structural', 'perturbed_structural'),
        ('Tree Edit', 'clean_tree_edit', 'perturbed_tree_edit'),
    ]
    
    for metric_name, clean_key, perturbed_key in metrics:
        if clean_key in per_sample and perturbed_key in per_sample:
            clean = per_sample[clean_key]
            perturbed = per_sample[perturbed_key]
            
            results[metric_name] = {
                'consistency': compute_consistency(clean, perturbed),
                'worst_case': compute_worst_case(clean, perturbed),
                'correlation': compute_correlation(clean, perturbed),
                'combined': compute_combined_robustness(clean, perturbed),
                'stats': {
                    'clean_mean': float(np.mean(clean)),
                    'perturbed_mean': float(np.mean(perturbed)),
                    'clean_std': float(np.std(clean)),
                    'perturbed_std': float(np.std(perturbed))
                }
            }
    
    return results
```

```python
def print_analysis(results):
    """
    Pretty print the analysis results.
    """
    if results is None:
        return
    
    print("\n" + "=" * 70)
    print(f"ALTERNATIVE ROBUSTNESS METRICS: {results['model'].upper()}")
    print(f"Samples: {results['samples']}, Perturbation Strength: {results['strength']}")
    print("=" * 70)
    
    for metric_name in ['CLIP', 'IoU', 'Structural', 'Tree Edit']:
        if metric_name not in results:
            continue
        
        m = results[metric_name]
        print(f"\n{'-' * 70}")
        print(f"{metric_name.upper()}")
        print(f"{'-' * 70}")
        
        # Basic stats
        stats = m['stats']
        print(f"  Mean: {stats['clean_mean']:.4f} (clean) -> {stats['perturbed_mean']:.4f} (perturbed)")
        print(f"  Std:  {stats['clean_std']:.4f} (clean) -> {stats['perturbed_std']:.4f} (perturbed)")
        
        # Consistency
        cons = m['consistency']
        print(f"\n  CONSISTENCY (higher=better, penalizes any change):")
        print(f"    Mean: {cons['mean']:.4f}")
        print(f"    Std:  {cons['std']:.4f}")
        print(f"    Min:  {cons['min']:.4f}")
        
        # Worst case
        wc = m['worst_case']
        print(f"\n  WORST-CASE:")
        print(f"    Max degradation:     {wc['max_degradation']:.4f} ({wc['max_degradation_pct']:.1f}%)")
        print(f"    P95 degradation:     {wc['p95_degradation']:.4f} ({wc['p95_degradation_pct']:.1f}%)")
        print(f"    Max absolute change: {wc['max_abs_change']:.4f}")
        print(f"    Samples degraded:    {wc['samples_degraded']}/{results['samples']}")
        print(f"    Samples improved:    {wc['samples_improved']}/{results['samples']}")
        
        # Correlation
        corr = m['correlation']
        if corr['spearman'] is not None:
            print(f"\n  CORRELATION (rank preservation):")
            print(f"    Spearman: {corr['spearman']:.4f} (p={corr['spearman_pvalue']:.4e})")
            print(f"    Pearson:  {corr['pearson']:.4f} (p={corr['pearson_pvalue']:.4e})")
        
        # Combined score
        comb = m['combined']
        rating = get_robustness_rating(comb['combined'])
        print(f"\n  COMBINED ROBUSTNESS SCORE: {comb['combined']:.4f} ({rating})")
        print(f"    Breakdown: Consistency={comb['consistency']:.3f}, Correlation={comb['correlation']:.3f}, ")
        print(f"               Mean Ratio={comb['mean_ratio']:.3f}, Worst-Case={comb['worst_case']:.3f}")
```

## Summary Comparison

```python
def print_summary_table(all_results):
    """
    Print a summary comparison table.
    """
    print("\n" + "=" * 100)
    print("SUMMARY: COMBINED ROBUSTNESS SCORES BY MODEL AND METRIC")
    print("=" * 100)
    print(f"{'Model':<20} {'CLIP':>12} {'IoU':>12} {'Structural':>12} {'Tree Edit':>12} {'Average':>12}")
    print("-" * 100)
    
    for results in all_results:
        if results is None:
            continue
        
        scores = []
        row = f"{results['model']:<20}"
        
        for metric_name in ['CLIP', 'IoU', 'Structural', 'Tree Edit']:
            if metric_name in results:
                score = results[metric_name]['combined']['combined']
                scores.append(score)
                row += f" {score:>11.4f}"
            else:
                row += f" {'N/A':>11}"
        
        if scores:
            avg = np.mean(scores)
            rating = get_robustness_rating(avg)
            row += f" {avg:>11.4f} ({rating})"
        
        print(row)
    
    print("\n" + "=" * 100)
    print("INTERPRETATION:")
    print("  >= 0.95: EXCELLENT - Nearly identical outputs regardless of perturbation")
    print("  >= 0.85: GOOD      - Minor variations, rankings preserved")
    print("  >= 0.70: MODERATE  - Noticeable changes but generally stable")
    print("  >= 0.50: POOR      - Significant instability")
    print("  <  0.50: UNRELIABLE - Outputs highly dependent on input quality")
    print("=" * 100)
```

```python
# Summary table
all_results = [qwen_results, websight_results, gemini_results, design2code_results]
print_summary_table(all_results)
```

## Consistency vs Degradation Comparison

```python
def compare_consistency_vs_degradation(all_results):
    """
    Compare consistency-based metric with current degradation-based metric.
    """
    print("\n" + "=" * 100)
    print("CONSISTENCY vs DEGRADATION COMPARISON")
    print("=" * 100)
    print("\nKey Insight: Consistency penalizes BOTH improvements and degradations.")
    print("If perturbed > clean, current metric shows 0% degradation but consistency < 1.0")
    print("\n" + "-" * 100)
    
    for results in all_results:
        if results is None:
            continue
        
        print(f"\n{results['model'].upper()}:")
        print(f"  {'Metric':<15} {'Clean Mean':>12} {'Pert Mean':>12} {'Change':>10} {'Consistency':>12} {'Improved?':>10}")
        
        for metric_name in ['CLIP', 'IoU', 'Structural']:
            if metric_name not in results:
                continue
            
            m = results[metric_name]
            clean_mean = m['stats']['clean_mean']
            pert_mean = m['stats']['perturbed_mean']
            change = pert_mean - clean_mean
            consistency = m['consistency']['mean']
            improved = "Yes" if change > 0.01 else ("No" if change < -0.01 else "Same")
            
            print(f"  {metric_name:<15} {clean_mean:>12.4f} {pert_mean:>12.4f} {change:>+10.4f} {consistency:>12.4f} {improved:>10}")

compare_consistency_vs_degradation(all_results)
```

---
## Worst-Case Analysis

```python
def worst_case_analysis(all_results):
    """
    Compare worst-case behavior across models.
    """
    print("\n" + "=" * 100)
    print("WORST-CASE ANALYSIS: MAXIMUM PER-SAMPLE DEGRADATION")
    print("=" * 100)
    print("\nFor safety-critical applications, worst-case matters more than average.")
    print("\n" + "-" * 100)
    
    for results in all_results:
        if results is None:
            continue
        
        print(f"\n{results['model'].upper()}:")
        print(f"  {'Metric':<15} {'Max Drop':>12} {'Max Drop %':>12} {'P95 Drop':>12} {'#Degraded':>12} {'#Improved':>12}")
        
        for metric_name in ['CLIP', 'IoU', 'Structural']:
            if metric_name not in results:
                continue
            
            wc = results[metric_name]['worst_case']
            n = results['samples']
            
            print(f"  {metric_name:<15} {wc['max_degradation']:>12.4f} {wc['max_degradation_pct']:>11.1f}% "
                  f"{wc['p95_degradation']:>12.4f} {wc['samples_degraded']:>11}/{n} {wc['samples_improved']:>11}/{n}")

worst_case_analysis(all_results)
```
